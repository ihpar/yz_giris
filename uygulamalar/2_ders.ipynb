```python
"""
öklid uzaklığı
v1: x1, y1, z1, ...
v2: x2, y2, z2, ...

öu = ((x1-x2)^2 + (y1-y2)^2 + (z1-z2)^2 + ...) ^ 1/2
"""

v1 = [1, 3, 2, 5, 0]
v2 = [0, 7, 1, 8, 15]

print(len(v1))

toplam = 0
for i in range(len(v1)):
    # print(v1[i], v2[i])
    fark = v1[i] - v2[i]
    karesi = fark ** 2
    toplam = toplam + karesi

uzaklik = toplam ** (1/2)
print(uzaklik)
```

```python
"""
cos sim
(v1 . v2) / (norm(v1) * norm(v2))
v1 = [x1, y1]
v2 = [x2, y2]
v1 . v2 = x1*x2 + y1*y2
norm(v1) = (x1**2 + y1**2) ** (1/2)
"""


def dot_prod(v1, v2):
    toplam = 0
    for i in range(len(v1)):
        carpim = v1[i] * v2[i]
        toplam = toplam + carpim

    return toplam


print(dot_prod([1, 2, 0], [3, 4, 5]))


def vec_norm(v):
    toplam = 0
    for bilesen in v:
        toplam = toplam + bilesen ** 2

    uzunluk = toplam ** (1/2)
    return uzunluk


print(vec_norm([3, 4]))
```

```python
def cos_sim(v1, v2):
    nokta_carpim = dot_prod(v1, v2)
    v1_uzunlugu = vec_norm(v1)
    v2_uzunlugu = vec_norm(v2)
    benzerlik = nokta_carpim / (v1_uzunlugu * v2_uzunlugu)
    return benzerlik


print(cos_sim([1, 7], [1, 7]))
print(cos_sim([1, 7], [2, 14]))
print(cos_sim([1, 7], [-1, -7]))
print(cos_sim([0, 7], [3, 0]))
```

```python
"""
numpy çözümü
"""

v1 = np.array([1, 7])
v2 = np.array([2, 14])

print(np.dot(v1, v2))

print(np.linalg.norm(v1))


def cos_sim_np(v1, v2):
    benzerlik = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
    return benzerlik


print(cos_sim_np(np.array([1, 7]), np.array([1, 7])))
print(cos_sim_np(np.array([1, 7]), np.array([2, 14])))
print(cos_sim_np(np.array([1, 7]), np.array([-1, -7])))
print(cos_sim_np(np.array([0, 7]), np.array([3, 0])))
```
